```python
import json
```

```python
# Load the DATA
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
print(data)
print(type(data))
```

```python
#clean and structure data
def clean_data(data):
    text_to_numbers={"one":1,"two":2,"three":3,"four":4,"five":5}
    clean_data= []
    unique_users = set()
    for user in data:
        raw_rating=str(user["rating"]).strip().lower()
        if (raw_rating in text_to_numbers):
            raw_rating=text_to_numbers[raw_rating]
        user["rating"]=raw_rating
        #Handle missing values
        raw_age=user.get("age")
        if(raw_age == None):
            user["age"]=None
        #deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        clean_data.append(user)
    return clean_data
```

```python
clean_data(data)
```
