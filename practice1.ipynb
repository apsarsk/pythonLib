```python
import json
```

```python
def read_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
        return data
```

```python
data=read_data("store_data.json")
```

```python
def clean_data(data):
    letter_to_rating={"one":1,"two":2,"three":3,"four":4,"five":5}
    clean_data=[]
    unique_data=set()
    
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if (raw_rating in letter_to_rating):
            raw_rating=letter_to_rating[raw_rating]
        user["rating"]=raw_rating
        #Handle mising values
        raw_age=user.get("age")
        if (raw_age==None):
            user["age"]=None
        #deduplication
        if (user["name"].strip() in unique_data):
            continue
        unique_data.add(user["name"])
        clean_data.append(user)
    return clean_data
            
            
        
        
            
```

```python
data=clean_data(data)
```

```python
#Get meanigful insight from data
def get_insights(data):
    total_rating=0
    for user in data:
        total_rating += float(user["rating"])
    print(f"avg rating ={total_rating/len(data)}")

#percentage of user with poor rating
    poor_rating=0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating += 1
    print(f"% of user  with poor rating = {poor_rating/len(data)*100}%")
            
```

```python
get_insights(data)
```

```python
def get_recommendations(data):
    recommendations=[]

    for user in data:
        curr_recommendation={}
        curr_recommendation["name"]=user["name"]
        if(float(user["rating"]) >=4 ):
            curr_recommendation['brand']="apple"
        else:
            curr_recommendation['brand']="sumsung"
        recommendations.append(curr_recommendation)
    return recommendations
```

```python
get_recommendations(data)
```
